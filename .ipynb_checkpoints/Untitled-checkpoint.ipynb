```python
# Cell 2 — imports and path
import os
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import linear_kernel

# Change path if your CSV is somewhere else
DATA_PATH = C:/Users/prate/Downloads/gift_dataset_50000.csv"  # or "gift_dataset_50000.csv"
print("Using data path:", DATA_PATH)
```

```python
# Cell 3 — load data with safe fallback
def load_data(path):
    if not os.path.exists(path):
        raise FileNotFoundError(f"File not found at {path}. Please place the CSV there or update DATA_PATH.")
    df = pd.read_csv(path)
    return df

df = load_data(DATA_PATH)
print("Loaded shape:", df.shape)
```
